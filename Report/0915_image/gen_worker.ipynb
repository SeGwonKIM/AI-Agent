# AI 이미지 생성 실습
계정도 키도 없이 이미지 한 장. 프롬프트를 URL에 실어 보내면 그림이 돌아온다.

[Pollinations.ai](https://pollinations.ai)의 무료 이미지 생성 API를 이용해,
문장(프롬프트)만으로 이미지를 생성하고 저장한 뒤 화면에 보여줍니다.

```python
# 필요한 라이브러리 확인 후, 없으면 설치 (구글 코랩 기준)
import importlib
import subprocess
import sys

try:
    import google.colab  # noqa: F401
    IN_COLAB = True
except ImportError:
    IN_COLAB = False

print("코랩 환경" if IN_COLAB else "코랩이 아닌 로컬/다른 환경")

required = {"requests": "requests", "PIL": "pillow", "matplotlib": "matplotlib"}

for module_name, pip_name in required.items():
    try:
        importlib.import_module(module_name)
    except ImportError:
        print(f"{pip_name} 설치 중...")
        subprocess.check_call([sys.executable, "-m", "pip", "install", "-q", pip_name])

print("라이브러리 준비 완료 (코랩은 대부분 기본 설치돼 있어 바로 통과됩니다)")
```

```python
# 계정도 키도 없이 이미지 한 장. 프롬프트를 URL에 실어 보내면 그림이 돌아온다.
import requests
from urllib.parse import quote

prompt = ("a single factory worker in a white safety helmet "
          "leaning over a stainless steel inspection line")

url = f"https://image.pollinations.ai/prompt/{quote(prompt)}"
r = requests.get(url, params={"width": 768, "height": 512, "nologo": "true"}, timeout=60)
r.raise_for_status()

IMAGE_PATH = "worker.jpg"
open(IMAGE_PATH, "wb").write(r.content)
print(f"{len(r.content):,} 바이트 저장")
```

## 생성된 이미지 확인
저장한 이미지를 바로 화면에 띄워서 결과를 확인합니다.

```python
from PIL import Image
import matplotlib.pyplot as plt

img = Image.open(IMAGE_PATH)

plt.imshow(img)
plt.axis("off")
plt.show()
```
