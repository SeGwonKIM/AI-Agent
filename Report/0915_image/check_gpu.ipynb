# GPU 확인
무엇이 붙었는지부터 확인한다. 배정되는 GPU는 그때그때 다르다.

Colab에서 **런타임 > 런타임 유형 변경 > 하드웨어 가속기: GPU** 로 설정한 뒤 실행하세요.

```python
# 무엇이 붙었는지부터 확인한다. 배정되는 GPU는 그때그때 다르다.
import torch

if not torch.cuda.is_available():
    raise SystemExit("GPU가 안 붙었습니다. 런타임 유형을 GPU로 바꾸고 다시 실행하세요.")

p = torch.cuda.get_device_properties(0)
major, minor = torch.cuda.get_device_capability(0)
print(f"{p.name} / VRAM {p.total_memory / 1024**3:.1f}GB / compute capability {major}.{minor}")
```
